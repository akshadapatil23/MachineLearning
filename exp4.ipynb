```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import r2_score
```

```python
import numpy as np
import pandas as pd

X = np.array([1,1,1,3,1,5,2,0,2,2,2,0,9,3,0,0,3,2,3,7,4,0,2,9])
Y = np.array([39000,45000,37700,39000,56000,60000,np.nan,np.nan,54000,54400,57100,32000,63200,56600])

X = X[:len(Y)]

df = pd.DataFrame({
    "Expression":X,
    "Salary":Y
})
print(df)
```

```python
plt.plot(X,Y, marker="D")
plt.xlabel("Expression")
plt.ylabel("Salary")
plt.title("Original Data - Experience vs Salary")
plt.grid()
plt.show()
```

```python
df.isnull()
```

```python
X = df[["Expression"]]
Y = df[["Salary"]]
X_train , X_test, Y_train, Y_test = train_test_split(X,Y,test_size = 0.2, random_state=0)
```

```python
X_test.shape
```

```python
Y_train.shape
```

```python
Y_test.shape
```

```python
X_train.shape
```

```python
model = LogisticRegression()
```

```python
Y_train_cleanned = Y_train.fillna(Y_train.mean())
model.fit(X_train,Y_train_cleanned)
```

```python
Y_pred = model.predict(X_test)
print(Y_pred)
```

```python
Y_test_cleaned = Y_test.fillna(Y_test.mean())
r2 = r2_score(Y_test_cleaned,Y_pred)
print(r2)
```

```python
exprience = np.array([[10]])
salary_pred = model.predict(exprience)
print(salary_pred)
```

```python
plt.plot(X,Y, label = "Actual Salary")
plt.plot(X,model.predict(X),label="Regression Line")
plt.xlabel("Expression")
plt.ylabel("Salary")
plt.title("original Data - Experience vs Salary")
plt.grid()
plt.legend()
plt.show()
```
